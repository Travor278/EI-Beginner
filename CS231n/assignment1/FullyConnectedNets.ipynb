Did you notice anything about the comparative difficulty of training the three-layer network vs. training the five-layer network? In particular, based on your experience, which network seemed more sensitive to the initialization scale? Why do you think that is the case?

$\color{blue}{\textit Your Answer:}$ In my experiments, the five-layer network was harder to train and clearly more sensitive to the initialization scale than the three-layer network. A practical reason is that in a deeper model, activations and gradients have to pass through more affine-ReLU blocks, so the effect of the initial weight scale compounds across layers. If the weights are too small, signals and gradients tend to shrink as they propagate; if they are too large, they can grow too quickly and make optimization unstable. Since the five-layer network has more repeated layers, this accumulation effect is stronger, so choosing a good initialization scale matters more than it does for the three-layer network.

## Inline Question 2:

AdaGrad, like Adam, is a per-parameter optimization method that uses the following update rule:

```
cache += dw**2
w += - learning_rate * dw / (np.sqrt(cache) + eps)
```

John notices that when he was training a network with AdaGrad that the updates became very small, and that his network was learning slowly. Using your knowledge of the AdaGrad update rule, why do you think the updates would become very small? Would Adam have the same issue?


$\color{blue}{\textit Your Answer:}$ In AdaGrad, the cache keeps accumulating $dw^2$, so it is monotone increasing over time. Since the update divides by $\sqrt{cache} + \epsilon$, the effective step size keeps shrinking as training goes on. This can make the updates very small and slow learning considerably, especially after many iterations. Adam is less prone to this issue because it does not use an ever-growing sum of squared gradients; instead, it uses exponentially decaying moving averages of both the first and second moments. That means older gradient information is gradually forgotten, so the denominator does not grow without bound in the same way, and the step sizes usually remain more stable.
